```python
import pandas as pd
import os
from glob import glob

folder = r"F:\开都河流域图绘制\cmip6模式数据整理 (1)\cmip6模式数据整理\wind\ssp126"

files = glob(os.path.join(folder, "*.xlsx"))

all_data = []

for file in files:
    try:
        if os.path.basename(file).startswith("~$"):
            continue
        
        df = pd.read_excel(file, header=None)
        
        # ❗ 如果第一行是路径（字符串），去掉
        if isinstance(df.iloc[0, 0], str):
            df = df.iloc[1:]
        
        # ❗ 全部转为数值
        df = df.apply(pd.to_numeric, errors='coerce')
        
        # ❗ 生成时间轴（这里一定要按你的真实数据改！）
        df["date"] = pd.date_range(start="2017-01-01", periods=len(df))
        
        all_data.append(df)
    
    except Exception as e:
        print(f"❌ 失败: {file}")
        print(e)

# 合并所有站点
data = pd.concat(all_data)

# 🎯 核心：按日期，对“每一列（模式）”求均值
daily_mean = data.groupby("date").mean()

# 保存
out_path = os.path.join(folder, "流域_各模式_日均值.xlsx")
daily_mean.to_excel(out_path)

print("✅ 完成！")
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# =========================
# 🎯 全局论文字体设置（关键）
# =========================
plt.rcParams.update({
    "font.family": "Times New Roman",
    "font.size": 14,
    "axes.labelsize": 16,
    "xtick.labelsize": 16,
    "ytick.labelsize": 16,
    "legend.fontsize": 12,
})

# =========================
# 路径
# =========================
root = r"D:\Python code\boshilunwen\vic\cmip"

valid_vars = ["pr", "tas", "tmax", "tmin", "wind"]

unit_map = {
    "pr": "mm",
    "tas": "℃",
    "tmax": "℃",
    "tmin": "℃",
    "wind": "m/s"
}

# 🎨 情景颜色（IPCC风格）
color_map = {
    "ssp126": "#1f77b4",
    "ssp245": "#2ca02c",
    "ssp370": "#ff7f0e",
    "ssp585": "#d62728"
}

# =========================
# 画布
# =========================
fig, axes = plt.subplots(5, 2, figsize=(14, 20))
letters = list("abcdefghij")

# =========================
# 主循环
# =========================
for i, var in enumerate(valid_vars):

    var_path = os.path.join(root, var)

    if not os.path.exists(var_path):
        print(f"⚠️ 缺少文件夹: {var}")
        continue

    ax_daily = axes[i, 0]
    ax_yearly = axes[i, 1]

    all_daily = []
    daily_clim_all = {}

    # =========================
    # 遍历情景
    # =========================
    for file in os.listdir(var_path):

        if file.endswith(".xlsx"):

            scen = file.replace(".xlsx", "")
            file_path = os.path.join(var_path, file)

            try:
                df = pd.read_excel(file_path)

                # 时间
                df.iloc[:, 0] = pd.to_datetime(df.iloc[:, 0])
                df["year"] = df.iloc[:, 0].dt.year

                # 多模式平均（日）
                df["mean"] = df.iloc[:, 1:-1].mean(axis=1)

                # 收集所有日数据（用于MME）
                all_daily.append(df["mean"])

                # =========================
                # 👉 日循环（每个情景）
                # =========================
                df_tmp = pd.DataFrame({
                    "date": df.iloc[:, 0],
                    "val": df["mean"]
                })
                df_tmp["doy"] = df_tmp["date"].dt.dayofyear

                clim_scen = df_tmp.groupby("doy")["val"].mean()
                daily_clim_all[scen] = clim_scen

                # =========================
                # 👉 年变化（含std）
                # =========================
                model_yearly = []

                for col in df.columns[1:-2]:
                    if var == "pr":
                        tmp = df.groupby("year")[col].sum()
                    else:
                        tmp = df.groupby("year")[col].mean()
                    model_yearly.append(tmp)

                model_yearly = pd.concat(model_yearly, axis=1)

                mean_y = model_yearly.mean(axis=1)
                std_y = model_yearly.std(axis=1)

                color = color_map.get(scen, "black")

                # 主线（MME）
                ax_yearly.plot(mean_y.index, mean_y.values,
                               color=color,
                               linewidth=1.8,
                               label=f"{scen}")

                # 阴影（±1σ）
                ax_yearly.fill_between(mean_y.index,
                                       mean_y - std_y,
                                       mean_y + std_y,
                                       color=color,
                                       alpha=0.2,
                                       label=f"{scen} ensemble spread" if i == 0 else None)

            except Exception as e:
                print(f"❌ 出错: {file_path}")
                print(e)

    # =========================
    # 👉 日尺度绘图
    # =========================
    for scen, clim_scen in daily_clim_all.items():
        color = color_map.get(scen, "gray")
        ax_daily.plot(clim_scen.index, clim_scen.values,
                      color=color,
                      linewidth=0.8,
                      alpha=0.4)

    # 👉 MME（日）
    if len(all_daily) > 0:
        daily_all = pd.concat(all_daily, axis=1)
        daily_mean = daily_all.mean(axis=1)

        df_tmp = pd.DataFrame({
            "date": pd.to_datetime(df.iloc[:, 0]),
            "val": daily_mean
        })
        df_tmp["doy"] = df_tmp["date"].dt.dayofyear

        clim = df_tmp.groupby("doy")["val"].mean()

        ax_daily.plot(clim.index, clim.values,
                      color="black",
                      linewidth=2.2,
                      label="MME")

    # =========================
    # 🎯 样式优化（无网格 + 外刻度）
    # =========================
    for ax in [ax_daily, ax_yearly]:
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.grid(False)  # ❗去网格
        ax.tick_params(direction='out')

    # 坐标轴
    ax_daily.set_xlabel("Day of Year")
    ax_yearly.set_xlabel("Year")

    ax_daily.set_ylabel(f"{var} ({unit_map.get(var, '')})")
    ax_yearly.set_ylabel(f"{var} ({unit_map.get(var, '')})")

    ax_daily.set_xlim(1, 365)
    
    ax_yearly.set_xlim(2017, 2070)

    # 子图编号
    ax_daily.text(0.02, 0.9, f"({letters[i*2]})",
                  transform=ax_daily.transAxes,
                  fontsize=13, fontweight='bold')

    ax_yearly.text(0.02, 0.9, f"({letters[i*2+1]})",
                   transform=ax_yearly.transAxes,
                   fontsize=13, fontweight='bold')

# =========================
# 🎯 图例（去重 + 合并）
# =========================
handles1, labels1 = axes[0, 0].get_legend_handles_labels()
handles2, labels2 = axes[0, 1].get_legend_handles_labels()

all_handles = handles1 + handles2
all_labels = labels1 + labels2

by_label = dict(zip(all_labels, all_handles))

fig.legend(by_label.values(),
           by_label.keys(),
           loc='upper center',
           ncol=5,
           frameon=False)

# =========================
# 布局
# =========================
plt.tight_layout(rect=[0, 0, 1, 0.96])

# 保存
save_path = os.path.join(root, "组图_最终论文版.png")
plt.savefig(save_path, dpi=300, bbox_inches='tight')

plt.show()

print("✅ 最终论文级图完成")
```

```python
import os
import pandas as pd
import numpy as np

# =========================
# 路径
# =========================
root = r"D:\Python code\boshilunwen\vic\cmip"

valid_vars = ["pr", "tas", "tmax", "tmin", "wind"]

result_all = []

# =========================
# 主循环
# =========================
for var in valid_vars:

    var_path = os.path.join(root, var)
    if not os.path.exists(var_path):
        continue

    for file in os.listdir(var_path):

        if not file.endswith(".xlsx"):
            continue

        scen = file.replace(".xlsx", "")
        file_path = os.path.join(var_path, file)

        try:
            df = pd.read_excel(file_path)

            # =========================
            # 时间处理
            # =========================
            df.iloc[:, 0] = pd.to_datetime(df.iloc[:, 0])
            df["year"] = df.iloc[:, 0].dt.year
            df["doy"] = df.iloc[:, 0].dt.dayofyear

            # 👉 模型列（关键）
            model_cols = df.columns[1:-2]

            # =========================
            # 年尺度
            # =========================
            model_yearly = []

            for col in model_cols:
                if var == "pr":
                    tmp = df.groupby("year")[col].sum()
                else:
                    tmp = df.groupby("year")[col].mean()
                model_yearly.append(tmp)

            model_yearly = pd.concat(model_yearly, axis=1)

            mean_y = model_yearly.mean(axis=1)
            std_y = model_yearly.std(axis=1)

            start = mean_y.loc[2020]
            end = mean_y.loc[2070]

            delta = end - start
            percent = delta / start * 100

            slope = np.polyfit(mean_y.index, mean_y.values, 1)[0]
            slope_decade = slope * 10

            interannual_std = mean_y.std()
            cv = interannual_std / mean_y.mean()

            uncertainty_mean = std_y.mean()
            uncertainty_max = std_y.max()

            # =========================
            # 日尺度（MME）
            # =========================
            df["mean"] = df[model_cols].mean(axis=1)

            clim = df.groupby("doy")["mean"].mean()

            max_val = clim.max()
            min_val = clim.min()
            amplitude = max_val - min_val

            max_day = clim.idxmax()
            min_day = clim.idxmin()

            daily_std_mean = clim.std()
            daily_cv = clim.std() / clim.mean()

            # =========================
            # 日STD（模型间）
            # =========================
            daily_std = df[model_cols].std(axis=1)

            clim_std = df.assign(std=daily_std).groupby("doy")["std"].mean()

            std_max = clim_std.max()
            std_max_day = clim_std.idxmax()
            std_mean = clim_std.mean()
            std_cv = clim_std.std() / clim_std.mean()

            # =========================
            # 汇总
            # =========================
            result_all.append({
                "var": var,
                "scenario": scen,

                # 年尺度
                "start_2020": start,
                "end_2070": end,
                "delta": delta,
                "percent_%": percent,
                "trend_per_year": slope,
                "trend_per_decade": slope_decade,
                "interannual_std": interannual_std,
                "cv": cv,
                "uncertainty_mean": uncertainty_mean,
                "uncertainty_max": uncertainty_max,

                # 日尺度
                "max": max_val,
                "min": min_val,
                "amplitude": amplitude,
                "max_doy": max_day,
                "min_doy": min_day,
                "daily_std_mean": daily_std_mean,
                "daily_cv": daily_cv,

                # 日STD
                "std_max": std_max,
                "std_max_doy": std_max_day,
                "std_mean": std_mean,
                "std_cv": std_cv
            })

        except Exception as e:
            print(f"❌ 出错: {file}")
            print(e)

# =========================
# 转DataFrame
# =========================
result_df = pd.DataFrame(result_all)

# =========================
# 中文表头
# =========================
result_df = result_df.rename(columns={

    "var": "变量",
    "scenario": "情景",

    # 年尺度
    "start_2020": "2020年值",
    "end_2070": "2070年值",
    "delta": "变化量",
    "percent_%": "变化率(%)",
    "trend_per_year": "趋势(每年)",
    "trend_per_decade": "趋势(每十年)",
    "interannual_std": "年际标准差",
    "cv": "年际变异系数",
    "uncertainty_mean": "不确定性均值",
    "uncertainty_max": "不确定性最大值",

    # 日尺度
    "max": "最大值",
    "min": "最小值",
    "amplitude": "振幅",
    "max_doy": "峰值日(DOY)",
    "min_doy": "谷值日(DOY)",
    "daily_std_mean": "日尺度标准差",
    "daily_cv": "日尺度变异系数",

    # 日STD
    "std_max": "STD最大值",
    "std_max_doy": "STD峰值日",
    "std_mean": "STD均值",
    "std_cv": "STD变异系数"
})

# =========================
# 输出
# =========================
save_path = os.path.join(root, "统计结果汇总_中文.xlsx")
result_df.to_excel(save_path, index=False)

print("✅ 完成：统计结果已输出 →", save_path)
```

```python
import os
import pandas as pd
import numpy as np

root = r"D:\Python code\boshilunwen\vic\cmip"
valid_vars = ["pr", "tas", "tmax", "tmin", "wind"]

for var in valid_vars:

    var_path = os.path.join(root, var)
    if not os.path.exists(var_path):
        continue

    # 输出文件
    yearly_path = os.path.join(root, f"{var}_年尺度.xlsx")
    daily_path = os.path.join(root, f"{var}_日尺度.xlsx")

    writer_year = pd.ExcelWriter(yearly_path, engine="openpyxl")
    writer_day = pd.ExcelWriter(daily_path, engine="openpyxl")

    for file in os.listdir(var_path):

        if not file.endswith(".xlsx"):
            continue

        scen = file.replace(".xlsx", "")
        file_path = os.path.join(var_path, file)

        try:
            df = pd.read_excel(file_path)

            df.iloc[:, 0] = pd.to_datetime(df.iloc[:, 0])
            df["year"] = df.iloc[:, 0].dt.year
            df["doy"] = df.iloc[:, 0].dt.dayofyear

            # 👉 模型列（关键）
            model_cols = df.columns[1:-2]

            # =========================
            # 年尺度（宽表）
            # =========================
            yearly_dict = {}

            for col in model_cols:
                if var == "pr":
                    tmp = df.groupby("year")[col].sum()
                else:
                    tmp = df.groupby("year")[col].mean()

                yearly_dict[col] = tmp

            yearly_df = pd.DataFrame(yearly_dict)

            # 加 MME
            yearly_df["MME"] = yearly_df.mean(axis=1)

            # index -> 列
            yearly_df = yearly_df.reset_index()
            yearly_df = yearly_df.rename(columns={"year": "Year"})

            # 保存
            yearly_df.to_excel(writer_year, sheet_name=scen, index=False)

            # =========================
            # 日尺度（宽表）
            # =========================
            daily_dict = {}

            for col in model_cols:
                clim = df.groupby("doy")[col].mean()
                daily_dict[col] = clim

            daily_df = pd.DataFrame(daily_dict)

            # 加 MME
            daily_df["MME"] = daily_df.mean(axis=1)

            daily_df = daily_df.reset_index()
            daily_df = daily_df.rename(columns={"doy": "DOY"})

            daily_df.to_excel(writer_day, sheet_name=scen, index=False)

        except Exception as e:
            print(f"❌ 出错: {file}")
            print(e)

    writer_year.close()
    writer_day.close()

    print(f"✅ {var} 完成")

print("🎉 全部导出完成（宽表，直接可作图）")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
from shapely.geometry import Point
import string
from matplotlib.colors import TwoSlopeNorm, Normalize
from matplotlib.ticker import FuncFormatter, FormatStrFormatter
from scipy.stats import linregress

# ==============================
# 1. 路径
# ==============================
station_file = r'D:\vic\station_data\station_info.txt'
base_path = r'D:\Python code\boshilunwen\vic\cmip'
shp_path = r'D:\oggm\流域划分\开都河\kd-ly.shp'

variables = ['pr', 'tas', 'tmax', 'tmin', 'wind']
scenarios = ['ssp126', 'ssp245', 'ssp370', 'ssp585']

unit_dict = {
    'pr': 'mm/year',
    'tas': '°C/year',
    'tmax': '°C/year',
    'tmin': '°C/year',
    'wind': 'm/s/year'
}

cmap_dict = {
    'pr': 'BrBG',
    'tas': 'coolwarm',
    'tmax': 'RdYlBu_r',
    'tmin': 'RdYlBu_r',
    'wind': 'PiYG'
}

def lon_formatter(x, pos): return f"{x:.1f}°E"
def lat_formatter(y, pos): return f"{y:.1f}°N"

# ==============================
# 站点
# ==============================
stations = pd.read_csv(station_file, sep=r'\s+')
stations.columns = ['station_id', 'lat', 'lon']
stations['station_id'] = stations['station_id'].astype(str)

# ==============================
# 流域
# ==============================
gdf = gpd.read_file(shp_path).to_crs(epsg=4326)

minx, miny, maxx, maxy = gdf.total_bounds
grid_lon = np.linspace(minx, maxx, 200)
grid_lat = np.linspace(miny, maxy, 200)
grid_lon, grid_lat = np.meshgrid(grid_lon, grid_lat)

# ==============================
# IDW
# ==============================
def idw(x, y, z, xi, yi, power=2):
    dist = np.sqrt((xi - x[:, None, None])**2 + (yi - y[:, None, None])**2)
    dist[dist == 0] = 1e-10
    weights = 1 / dist**power
    return np.sum(weights * z[:, None, None], axis=0) / np.sum(weights, axis=0)

points = np.vstack((grid_lon.ravel(), grid_lat.ravel())).T
mask = np.array([gdf.unary_union.contains(Point(xy)) for xy in points]).reshape(grid_lon.shape)

# ==============================
# 趋势
# ==============================
def calc_trend(df):
    if 'year' not in df.columns:
        df['year'] = np.arange(len(df))
    years = df['year'].values
    vals = df['multi_mean'].values
    if len(vals) < 5 or np.all(np.isnan(vals)):
        return np.nan, np.nan
    slope, _, _, p, _ = linregress(years, vals)
    return slope, p

# ==============================
# 绘图
# ==============================
fig, axes = plt.subplots(len(variables), len(scenarios),
                         figsize=(18, 9), sharex=True, sharey=True)

label_index = 0

for i, var in enumerate(variables):
    temp_results = {}
    all_slopes = []

    for scen in scenarios:
        data_path = os.path.join(base_path, var, scen)
        records = []
        sig_points = []

        for j in range(len(stations)):
            sid = stations.loc[j, 'station_id']
            file_path = os.path.join(data_path, f"{sid}.csv")

            if not os.path.exists(file_path):
                continue

            df = pd.read_csv(file_path)
            slope, p = calc_trend(df)

            if np.isnan(slope):
                continue

            lon = float(stations.loc[j, 'lon'])
            lat = float(stations.loc[j, 'lat'])

            records.append([lon, lat, slope, p])

            if p < 0.05:
                sig_points.append([lon, lat])

        if len(records) == 0:
            temp_results[scen] = (np.full(grid_lon.shape, np.nan), [], [], [], [])
            all_slopes.append(np.full(grid_lon.shape, np.nan))
            continue

        records = np.array(records)
        lons, lats = records[:, 0], records[:, 1]
        slopes = records[:, 2]
        pvals = records[:, 3]

        grid_slope = idw(lons, lats, slopes, grid_lon, grid_lat)
        grid_slope = np.where(mask, grid_slope, np.nan)

        temp_results[scen] = (grid_slope, lons, lats, pvals, sig_points)
        all_slopes.append(grid_slope)

    if i not in [1, 2, 3]:
        vmin_global = np.nanmin(all_slopes)
        vmax_global = np.nanmax(all_slopes)
        if vmin_global < 0 and vmax_global > 0:
            norm_global = TwoSlopeNorm(vmin=vmin_global, vcenter=0, vmax=vmax_global)
        else:
            norm_global = Normalize(vmin=vmin_global, vmax=vmax_global)

    for j, scen in enumerate(scenarios):
        ax = axes[i, j]

        if scen not in temp_results:
            ax.axis('off')
            continue

        grid_slope, lons, lats, pvals, sig_points = temp_results[scen]

        if i in [1, 2, 3]:
            vmin_sub = np.nanmin(grid_slope)
            vmax_sub = np.nanmax(grid_slope)
            if vmin_sub < 0 and vmax_sub > 0:
                norm_sub = TwoSlopeNorm(vmin=vmin_sub, vcenter=0, vmax=vmax_sub)
            else:
                norm_sub = Normalize(vmin=vmin_sub, vmax=vmax_sub)

            contour = ax.contourf(grid_lon, grid_lat, grid_slope,
                                  levels=15, cmap=cmap_dict[var], norm=norm_sub)
        else:
            contour = ax.contourf(grid_lon, grid_lat, grid_slope,
                                  levels=15, cmap=cmap_dict[var], norm=norm_global)

        gdf.boundary.plot(ax=ax, color='black', linewidth=0.5)

        # 去边框
        for spine in ax.spines.values():
            spine.set_visible(False)

        # 显著性
        if sig_points:
            sig_lons, sig_lats = zip(*sig_points)
            ax.scatter(sig_lons, sig_lats, s=15, c='black', alpha=0.8, zorder=5)

        # 标题（压缩间距）
        if i == 0:
            ax.set_title(scen.upper(), fontsize=16, fontweight='bold', pad=2)

        # 行标签
        if j == 0:
            ax.set_ylabel(f"{var} ({unit_dict[var]})", fontsize=14)

        # 经纬度
        if i == len(variables) - 1:
            ax.xaxis.set_major_formatter(FuncFormatter(lon_formatter))
            ax.tick_params(axis='x', labelsize=11)
        else:
            ax.tick_params(axis='x', labelbottom=False)

        if j == 0:
            ax.yaxis.set_major_formatter(FuncFormatter(lat_formatter))
            ax.tick_params(axis='y', labelsize=11)
        else:
            ax.tick_params(axis='y', labelleft=False)

        # 子图编号
        ax.text(0.02, 0.95,
                f"({string.ascii_lowercase[label_index]})",
                transform=ax.transAxes,
                fontsize=11, fontweight='bold', va='top')

        label_index += 1

        # 更短的 colorbar
        cbar = fig.colorbar(contour, ax=ax,
                            fraction=0.02,
                            pad=0.015,
                            shrink=0.6,
                            aspect=30)
        cbar.ax.tick_params(labelsize=10)

        if i in [1, 2, 3]:
            cbar.ax.yaxis.set_major_formatter(FormatStrFormatter('%.4f'))

# ==============================
# 布局（关键压缩）
# ==============================
plt.subplots_adjust(left=0.07, right=0.92,
                    bottom=0.06, top=0.96,
                    wspace=0.25, hspace=0.005)

# ==============================
# 保存
# ==============================
plt.savefig(
    r'D:\Python code\boshilunwen\vic\cmip\trend_significance_each_cbar_mono.png',
    dpi=300,
    bbox_inches='tight'  
)

plt.show()
```
